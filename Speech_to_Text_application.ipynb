```python
# ============================================================
# SPEECH-TO-TEXT APPLICATION
# Google Colab - Single Cell
# ============================================================

!pip -q install openai-whisper gradio

import whisper
import gradio as gr

# ------------------------------------------------------------
# Load Whisper Model
# ------------------------------------------------------------

print("Loading Whisper model...")

model = whisper.load_model("base")

print("Whisper model loaded successfully!")


# ------------------------------------------------------------
# Speech-to-Text Function
# ------------------------------------------------------------

def speech_to_text(audio):

    if audio is None:
        return "Please upload an audio file."

    try:
        result = model.transcribe(audio)

        text = result["text"]

        return text.strip()

    except Exception as e:
        return f"Error: {str(e)}"


# ------------------------------------------------------------
# Gradio Application
# ------------------------------------------------------------

with gr.Blocks(title="Speech-to-Text Application") as app:

    gr.Markdown(
        """
        # 🎤 Speech-to-Text Application

        ### Automatic Speech Recognition using Whisper

        Upload an audio file and convert your speech into text
        automatically.
        """
    )

    audio_input = gr.Audio(
        sources=["upload", "microphone"],
        type="filepath",
        label="Upload or Record Audio"
    )

    convert_button = gr.Button(
        "🎙️ Convert Speech to Text",
        variant="primary"
    )

    text_output = gr.Textbox(
        label="Transcribed Text",
        lines=10,
        placeholder="Your converted text will appear here..."
    )

    convert_button.click(
        fn=speech_to_text,
        inputs=audio_input,
        outputs=text_output
    )

    gr.Markdown(
        """
        **Technologies:** Python | Whisper | Speech Recognition | Gradio | Google Colab
        """
    )


# ------------------------------------------------------------
# Launch Application
# ------------------------------------------------------------

app.launch(share=True)
```
